```python
# Task 2: Mark the success/failed launches for each site on the map
# Color function based on launch outcome
def assign_marker_color(launch_outcome):
    if launch_outcome == 1:
        return 'green'
    else:
        return 'red'

# Add markers for each launch
for index, record in spacex_df.iterrows():
    
    # Assign color based on outcome
    marker_color = assign_marker_color(record['class'])
    
    # Add circle marker
    folium.Circle(
        location=[record['Lat'], record['Long']],
        radius=100,
        color=marker_color,
        fill=True,
        fill_color=marker_color,
        fill_opacity=0.7,
        popup=folium.Popup(
            f"Launch Site: {record['Launch Site']}<br>"
            f"Outcome: {'Success' if record['class'] == 1 else 'Failed'}",
            max_width=200
        )
    ).add_to(site_map)
    
    # Add marker with icon
    folium.Marker(
        location=[record['Lat'], record['Long']],
        popup=folium.Popup(
            f"Launch Site: {record['Launch Site']}<br>"
            f"Outcome: {'Success' if record['class'] == 1 else 'Failed'}",
            max_width=200
        ),
        icon=folium.Icon(
            color=marker_color,
            icon='ok-sign' if record['class'] == 1 else 'remove-sign',
            prefix='glyphicon'
        )
    ).add_to(site_map)

# Display the map
site_map
```

```python
# TASK 3: Calculate the distances between a launch site to its proximities
from math import sin, cos, sqrt, atan2, radians

# Function to calculate distance between two coordinates using Haversine formula
def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6373.0  # radius of Earth in kilometers
    
    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)
    
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    
    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    
    distance = R * c
    return distance

# Coordinates of launch site (e.g. KSC LC-39A)
launch_site_lat = 28.573255
launch_site_lon = -80.646895

# Coordinates of proximities
proximities = {
    'Coastline':   [28.56367, -80.57163],
    'City':        [28.10473, -80.64531],
    'Highway':     [28.56358, -80.57101],
    'Railway':     [28.57206, -80.58526]
}

# Calculate and display distances
for proximity, coords in proximities.items():
    distance = calculate_distance(
        launch_site_lat, launch_site_lon,
        coords[0], coords[1]
    )
    print(f"Distance from launch site to {proximity}: {distance:.2f} km")
    
    # Add a marker for each proximity
    folium.Marker(
        location=coords,
        popup=folium.Popup(
            f"{proximity}<br>Distance: {distance:.2f} km",
            max_width=200
        ),
        icon=folium.Icon(color='blue', icon='info-sign')
    ).add_to(site_map)
    
    # Draw a line between launch site and proximity
    folium.PolyLine(
        locations=[
            [launch_site_lat, launch_site_lon],
            coords
        ],
        weight=2,
        color='blue',
        opacity=0.8,
        popup=f"{proximity}: {distance:.2f} km"
    ).add_to(site_map)

site_map
```
